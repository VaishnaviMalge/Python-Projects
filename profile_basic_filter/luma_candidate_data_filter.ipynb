```python
import pandas as pd
import os
```

```python
luma_file = r"C:\Users\User\Downloads\reviews_Agentic_Engineering_Multi_Agent_Systems_Talks_AI_Agent_Workshops_Ft_Snowflake_Cognee.csv"

data = pd.read_csv(luma_file)

file_name = os.path.basename(luma_file)
file_name
```

```python
data1= data[['name','linkedin', 'profile_review','relevant_profile','job_title_last_company', 'job_title_last_5_years', 
        'location','is_open_to_work', 'is_engaged_in_job_market', 'recent_job_change']]
```

```python
# make format consistant to avoid any error, remove leading trailing space, capitalization

cols = [
    "profile_review",
    "relevant_profile",
    "job_title_last_company",
    "job_title_last_5_years"
]

data1[cols] = data1[cols].apply(lambda x: x.str.strip().str.title())
```

```python
# filter out parameters

desired_profile_review = ["Good", "Elite"]
desired_relevant_profile_status = ["Yes"]

# list of role that we do not want
non_target_roles = [
    "advocate",
    "Advisor",
    "Analyst",
    "architect",
    "blockchain engineer",
    "Consultant",
    "creative lead",
    "Cybersecurity",
    "data engineer",
    "data product",
    "developer relations",
    "devrel",
    "director of engineering",
    "head of data",
    "just some guy",
    "Marketing",
    "Owner",
    "product",
    "professor",
    "program manager",
    "reliability engineer",
    "Solutions Engineer",
    "technical project manager"
]
```

```python
# filter out byt profile_review 

data2 = data1[data1["profile_review"].isin(desired_profile_review)]                         # profile_review = Good, Elite

                
# filter out by relevant profile
data3 = data2[data2["relevant_profile"].isin(desired_relevant_profile_status)]              # relevant_profile = yes
```

```python
data3.head()
```

```python
# substring matching

pattern = "|".join(role.lower() for role in non_target_roles)           # '|' --> OR; stores lowercase form of all non_target_roles in or ex. A|B|C


# remove: boolean; if any of the column has non desired target role then it mark as true


remove = (
    data3["job_title_last_company"].str.lower().str.contains(pattern, na=False) |       # na=False-->if value=NaN treat as not a match don't give error
    data3["job_title_last_5_years"].str.lower().str.contains(pattern, na=False)
)

 
```

```python
# In case you want to check data that has been removed
# data3[remove]             gives complete data

# gte list of titles removed
removed_titles = data3[remove][[
    "job_title_last_company",
    "job_title_last_5_years"
]]

removed_titles
```

```python

# filtered data

filtered_data = data3[remove == False]           # filter out all non non_target_roles 
                                                 # == False  keep the ones that don't contain those non target roles.
 
# check total filtered candidates
print("Total candidates:", len(filtered_data))
```

# Candidate Filter

## Overview

This filtering process identifies relevant profiles based on **Profile Review** (`Good`, `Elite`) and **Relevant Profile** status, while also filtering out non-target job roles using pattern matching. It allows you to see which roles were removed, track the total number of profiles filtered, validate the filtering results, and download the final filtered dataset for further use.



### Handling Job Title Variations with Pattern Matching

Initially, I used exact-match filtering, which required every job-title variation to be added separately. To overcome this limitation, I used substring/pattern matching so that a role such as `"Data Engineer"` also matches variations like `"Senior Data Engineer"`, `"Junior Data Engineer"`, and `"Data Engineer II"` without manually adding each variation.

### Boolean Mask for Testing

Instead of directly filtering the dataset, I created a `remove` Boolean mask that marks matching profiles as `True`. This allows me to create the final filtered dataset while also checking the profiles marked for removal to verify that the correct roles are being filtered.
